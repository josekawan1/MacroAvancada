This routine will serve me as a guide to future time series analysis. This activity flow is based on the Box-Jenkins methodology, the classes of advanced macroeconomics course that i take at my master and my experience.

- **Plot the serie**
- **Identification**
    - Plot ACF and PACF
    - Autoarima, Loop with arima or trial and error
  
***Observation***: i will be analysing time series with the *out-of-sample* way, with train and test data. Therefore, *i have to run autoarima, loop or trial and error with my train data*, not with the full sample. There is no rule of thumb to define the train size and the test size (75%, 80%, 85%...). The minimum test size should be the number of observations the will be predicted after. Example: if i wanna predict the next 12 months of inflation, my test data should have at least 12 observations.

With the best models in hand - based on AIC, BIC and Hannah Queen information criteria - i have to estimate theses models

- **Estimation**
- **Test period forecast**

With the estimated models i ran, i will forecast the values of the test period and get the fit statistics (MAE, RMSE, MSE, Diebold-Mariano). NOTE: here i do not have to check residuals of the model again, just the residuals of the forecast. With the best models based in this fit statistics i will forecast the next periods of the serie.

- **Forecast**

The final step is to predict the n-ahead values of the serie.
